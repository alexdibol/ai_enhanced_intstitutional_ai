```python
from pathlib import Path
import sys, json
# Locate package root whether running locally or in Colab after upload/unzip
HERE=Path.cwd().resolve()
ROOT=next((p for p in [HERE,*HERE.parents] if (p/'04_Source_Code/exobrain.py').exists()), None)
if ROOT is None:
    raise FileNotFoundError('Open the notebook from inside the extracted reference package.')
sys.path.insert(0,str(ROOT/'04_Source_Code'))
import exobrain as ex
CORPUS=ROOT/'03_Synthetic_Corpus'
OUT=ROOT/'05_Outputs'
OUT.mkdir(exist_ok=True)
print('Project root:',ROOT)
```

## Discover files

```python
files=ex.discover(CORPUS)
[str(p.relative_to(CORPUS)) for p in files]
```

## Create discovery report

```python
report={'corpus_path':str(CORPUS),'artifact_count':len(files),'paths':[str(p.relative_to(CORPUS)) for p in files]}
ex.write_json(OUT/'NB04_discovery_report.json',report)
assert report['artifact_count']>0
report
```
