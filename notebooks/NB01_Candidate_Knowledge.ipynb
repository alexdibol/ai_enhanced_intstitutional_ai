**Objective:** Convert corpus artifacts into candidate knowledge records.

Part of **The ExoBrain Project: Engineering Human Cognitive Operating Systems in the Age of Artificial Intelligence**.

```python
from pathlib import Path
import sys, json
# Locate package root whether running locally or in Colab after upload/unzip
HERE=Path.cwd().resolve()
ROOT=next((p for p in [HERE,*HERE.parents] if (p/'04_Source_Code/exobrain.py').exists()), None)
if ROOT is None:
    raise FileNotFoundError('Open the notebook from inside the extracted reference package.')
sys.path.insert(0,str(ROOT/'04_Source_Code'))
import exobrain as ex
CORPUS=ROOT/'03_Synthetic_Corpus'
OUT=ROOT/'05_Outputs'
OUT.mkdir(exist_ok=True)
print('Project root:',ROOT)
```

## Discover source artifacts

```python
inventory=ex.build_inventory(CORPUS)
print('Artifacts:',len(inventory))
inventory[:2]
```

## Create candidates

```python
candidates=[ex.candidate_from_artifact(a) for a in inventory]
ex.write_json(OUT/'NB01_candidate_knowledge.json',candidates)
candidates[:2]
```

## Validate candidate structure

```python
report=ex.validate_candidates(candidates)
ex.write_json(OUT/'NB01_candidate_validation.json',report)
assert all(x['valid'] for x in report)
report[:3]
```
