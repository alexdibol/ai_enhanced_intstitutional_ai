**Objective:** Construct the governed authoritative Research Corpus manifest.

Part of **The ExoBrain Project: Engineering Human Cognitive Operating Systems in the Age of Artificial Intelligence**.

```python
from pathlib import Path
import sys, json
# Locate package root whether running locally or in Colab after upload/unzip
HERE=Path.cwd().resolve()
ROOT=next((p for p in [HERE,*HERE.parents] if (p/'04_Source_Code/exobrain.py').exists()), None)
if ROOT is None:
    raise FileNotFoundError('Open the notebook from inside the extracted reference package.')
sys.path.insert(0,str(ROOT/'04_Source_Code'))
import exobrain as ex
CORPUS=ROOT/'03_Synthetic_Corpus'
OUT=ROOT/'05_Outputs'
OUT.mkdir(exist_ok=True)
print('Project root:',ROOT)
```

## Load artifacts and approvals

```python
inventory=ex.read_json(OUT/'NB05_artifact_inventory.json') if (OUT/'NB05_artifact_inventory.json').exists() else ex.build_inventory(CORPUS)
reviewed=ex.read_json(OUT/'NB03_human_review_log.json') if (OUT/'NB03_human_review_log.json').exists() else [ex.review_candidate(ex.candidate_from_artifact(a)) for a in inventory]
len(inventory),len(reviewed)
```

## Build manifest

```python
manifest=ex.corpus_manifest(inventory,reviewed)
ex.write_json(OUT/'NB06_corpus_manifest.json',manifest)
manifest['corpus_id'],manifest['artifact_count'],manifest['fingerprint'][:16]
```

## Confirm governed admission

```python
assert manifest['artifact_count']==sum(x['status']=='approved' for x in reviewed)
{'admitted':manifest['artifact_count'],'approved':sum(x['status']=='approved' for x in reviewed)}
```
