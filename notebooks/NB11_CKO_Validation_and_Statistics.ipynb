# NB11 - CKO Validation and Statistics

**Objective:** Validate CKOs and produce knowledge-system statistics.

Part of **The ExoBrain Project: Engineering Human Cognitive Operating Systems in the Age of Artificial Intelligence**.

```python
from pathlib import Path
import sys, json
# Locate package root whether running locally or in Colab after upload/unzip
HERE=Path.cwd().resolve()
ROOT=next((p for p in [HERE,*HERE.parents] if (p/'04_Source_Code/exobrain.py').exists()), None)
if ROOT is None:
    raise FileNotFoundError('Open the notebook from inside the extracted reference package.')
sys.path.insert(0,str(ROOT/'04_Source_Code'))
import exobrain as ex
CORPUS=ROOT/'03_Synthetic_Corpus'
OUT=ROOT/'05_Outputs'
OUT.mkdir(exist_ok=True)
print('Project root:',ROOT)
```

## Load registry

```python
registry=ex.read_json(OUT/'NB10_cko_registry.json')
ckos=registry['ckos']; relationships=registry['relationships']
```

## Validate CKOs

```python
validation=ex.validate_ckos(ckos)
ex.write_json(OUT/'NB11_cko_validation.json',validation)
assert validation['valid_count']==validation['total'] and validation['unique_ids'] and validation['all_traceable']
validation
```

## Generate statistics

```python
stats={'cko_count':len(ckos),'relationship_count':len(relationships),'source_artifact_count':len({c['source_artifact_id'] for c in ckos}),'average_units_per_artifact':round(len(ckos)/len({c['source_artifact_id'] for c in ckos}),2)}
ex.write_json(OUT/'NB11_knowledge_statistics.json',stats)
stats
```
