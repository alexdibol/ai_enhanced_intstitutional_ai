**Objective:** Extract normalized metadata and integrity hashes from discovered artifacts.

Part of **The ExoBrain Project: Engineering Human Cognitive Operating Systems in the Age of Artificial Intelligence**.

```python
from pathlib import Path
import sys, json
# Locate package root whether running locally or in Colab after upload/unzip
HERE=Path.cwd().resolve()
ROOT=next((p for p in [HERE,*HERE.parents] if (p/'04_Source_Code/exobrain.py').exists()), None)
if ROOT is None:
    raise FileNotFoundError('Open the notebook from inside the extracted reference package.')
sys.path.insert(0,str(ROOT/'04_Source_Code'))
import exobrain as ex
CORPUS=ROOT/'03_Synthetic_Corpus'
OUT=ROOT/'05_Outputs'
OUT.mkdir(exist_ok=True)
print('Project root:',ROOT)
```

## Build metadata inventory

```python
inventory=ex.build_inventory(CORPUS)
ex.write_json(OUT/'NB05_artifact_inventory.json',inventory)
ex.write_inventory_csv(OUT/'NB05_artifact_inventory.csv',inventory)
inventory[:2]
```

## Inspect quality

```python
quality={'count':len(inventory),'all_hashed':all(len(x['sha256'])==64 for x in inventory),'unique_ids':len({x['artifact_id'] for x in inventory})==len(inventory)}
ex.write_json(OUT/'NB05_metadata_quality.json',quality)
assert all(quality.values()) if isinstance(quality.get('count'),bool) else quality['all_hashed'] and quality['unique_ids']
quality
```
