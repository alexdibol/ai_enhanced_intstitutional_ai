# NB02 - Governance Pipeline

**Objective:** Apply explicit governance gates to candidate knowledge.

Part of **The ExoBrain Project: Engineering Human Cognitive Operating Systems in the Age of Artificial Intelligence**.

```python
from pathlib import Path
import sys, json
# Locate package root whether running locally or in Colab after upload/unzip
HERE=Path.cwd().resolve()
ROOT=next((p for p in [HERE,*HERE.parents] if (p/'04_Source_Code/exobrain.py').exists()), None)
if ROOT is None:
    raise FileNotFoundError('Open the notebook from inside the extracted reference package.')
sys.path.insert(0,str(ROOT/'04_Source_Code'))
import exobrain as ex
CORPUS=ROOT/'03_Synthetic_Corpus'
OUT=ROOT/'05_Outputs'
OUT.mkdir(exist_ok=True)
print('Project root:',ROOT)
```

## Load or create candidates

```python
p=OUT/'NB01_candidate_knowledge.json'
if p.exists(): candidates=ex.read_json(p)
else: candidates=[ex.candidate_from_artifact(a) for a in ex.build_inventory(CORPUS)]
len(candidates)
```

## Apply policy checks

```python
policy=[]
for c in candidates:
    ok=bool(c['title'] and c['provenance'].get('sha256'))
    policy.append({'candidate_id':c['candidate_id'],'policy_pass':ok,'rule':'title-and-provenance-required'})
ex.write_json(OUT/'NB02_policy_checks.json',policy)
assert all(x['policy_pass'] for x in policy)
policy[:3]
```

## Build governance queue

```python
queue=[dict(c,governance_gate='eligible_for_human_review') for c in candidates]
ex.write_json(OUT/'NB02_governance_queue.json',queue)
queue[:2]
```
