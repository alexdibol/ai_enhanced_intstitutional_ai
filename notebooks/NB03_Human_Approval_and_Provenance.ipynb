**Objective:** Simulate auditable human admission decisions and preserve review history.

Part of **The ExoBrain Project: Engineering Human Cognitive Operating Systems in the Age of Artificial Intelligence**.

```python
from pathlib import Path
import sys, json
# Locate package root whether running locally or in Colab after upload/unzip
HERE=Path.cwd().resolve()
ROOT=next((p for p in [HERE,*HERE.parents] if (p/'04_Source_Code/exobrain.py').exists()), None)
if ROOT is None:
    raise FileNotFoundError('Open the notebook from inside the extracted reference package.')
sys.path.insert(0,str(ROOT/'04_Source_Code'))
import exobrain as ex
CORPUS=ROOT/'03_Synthetic_Corpus'
OUT=ROOT/'05_Outputs'
OUT.mkdir(exist_ok=True)
print('Project root:',ROOT)
```

## Load governance queue

```python
p=OUT/'NB02_governance_queue.json'
queue=ex.read_json(p) if p.exists() else [ex.candidate_from_artifact(a) for a in ex.build_inventory(CORPUS)]
len(queue)
```

## Record human decisions

```python
reviewed=[ex.review_candidate(c) for c in queue]
ex.write_json(OUT/'NB03_human_review_log.json',reviewed)
reviewed[:2]
```

## Summarize approvals

```python
summary={'approved':sum(x['status']=='approved' for x in reviewed),'rejected':sum(x['status']=='rejected' for x in reviewed),'total':len(reviewed)}
ex.write_json(OUT/'NB03_approval_summary.json',summary)
assert summary['approved']==summary['total']
summary
```
