Part of **The ExoBrain Project: Engineering Human Cognitive Operating Systems in the Age of Artificial Intelligence**.

```python
from pathlib import Path
import sys, json
# Locate package root whether running locally or in Colab after upload/unzip
HERE=Path.cwd().resolve()
ROOT=next((p for p in [HERE,*HERE.parents] if (p/'04_Source_Code/exobrain.py').exists()), None)
if ROOT is None:
    raise FileNotFoundError('Open the notebook from inside the extracted reference package.')
sys.path.insert(0,str(ROOT/'04_Source_Code'))
import exobrain as ex
CORPUS=ROOT/'03_Synthetic_Corpus'
OUT=ROOT/'05_Outputs'
OUT.mkdir(exist_ok=True)
print('Project root:',ROOT)
```

## Load manifest

```python
manifest=ex.read_json(OUT/'NB06_corpus_manifest.json') if (OUT/'NB06_corpus_manifest.json').exists() else ex.corpus_manifest(ex.build_inventory(CORPUS),[ex.review_candidate(ex.candidate_from_artifact(a)) for a in ex.build_inventory(CORPUS)])
```

## Verify source integrity

```python
integrity=ex.integrity_report(CORPUS,manifest)
ex.write_json(OUT/'NB07_integrity_report.json',integrity)
assert integrity['all_valid']
integrity['all_valid'],len(integrity['checks'])
```

## Detect duplicates

```python
dups=ex.duplicate_groups(manifest['artifacts'])
ex.write_json(OUT/'NB07_duplicate_report.json',dups)
{'duplicate_groups':len(dups)}
```
